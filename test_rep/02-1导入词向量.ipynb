```python
from gensim.models import KeyedVectors
model_path = './data/sgns.weibo.word'
model = KeyedVectors.load_word2vec_format(model_path)
```

```python
print(model.vector_size)  # 词向量维度
print(model['地铁'])       # 查看某个词的词向量
```

```python
similarity = model.similarity('上海市', '城市')  # 查看两个词的相似度
print(similarity)
```

```python
len(model.index_to_key)  # 词汇表大小,有多少个词
```

```python
model.most_similar(positive=['男人','女孩'],negative=['男孩'],topn = 3)   # 查看最相似的词
```
