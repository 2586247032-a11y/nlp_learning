```python
from torch import nn
from gensim.models import KeyedVectors
import torch
import jieba
```

```python
# 1.加载词向量
wv = KeyedVectors.load_word2vec_format('./data/word2vec.txt')
```

```python
# 2.处理OOV
unk_token = "<unk>"
index2word = [unk_token] + wv.index_to_key    # 列表形式,第一个元素是unk_token
word2index = {token: index for index, token in enumerate(index2word)}   # 字典形式,每个元素是单词和它对应的索引{'词1':0,'词2':1,...}  注意要反过来
```

```python
# 3.准备词向量矩阵
# 先定义
num_embeggings = len(index2word)   # 词向量个数
embedding_dim = wv.vector_size          # 词向量维度
embedding_matrix = torch.randn(num_embeggings, embedding_dim)
# 再填充
for index,word in enumerate(index2word):  # 遍历词向量     index_to_key是列表形式,每个元素是词向量对应的单词['词1','词2',...]
    if word in wv:
        embedding_matrix[index] = torch.tensor(wv[word])  # 填充词向量   ,wv[word]是词向量对应的单词的词向量    unk保持原始的随机生成的向量，其余的把训练好的填充进去
```

```python
# 4.定义Embedding层
embedding = nn.Embedding.from_pretrained(embedding_matrix,freeze=True)   # 冻结词向量,默认不更新
```

```python
# 5.测试
text = '我喜欢乘坐宇宙飞船'
tokens = jieba.lcut(text)
input_ids = [word2index.get(token, word2index[unk_token]) for token in tokens]    # key_to_index是字典,内容是每个词和它对应的索引{’词1’:索引,’词2’:索引,…}    如果找不到的话就返回0索引（表示是unk_token）   input_ids是个列表，里面装的是分完词后每个词的词向量
input_tensor = torch.tensor(input_ids)    # 把输入改成张量的形式
embedding(input_tensor)
```
